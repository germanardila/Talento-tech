```python
#llamar a panda
import pandas as pd
datos1 =[1,2,2,3,4,4,5,6]
datos2 =[2,3,4,4,4,6,5,7]

#Crear diccionario
Data={"x":datos1,"y":datos2}

# crear la tabla con panda
df=pd.DataFrame(Data)
df
#crear la columna xy
df["xy"] = df["x"] *df['y']

#crear la columna x2
df["x2"] = df['x'] **2

df
#SUMA DE TOTALES
df.loc[ "suma"] =df.sum()
df
```

```python

# CREACION DE TOTALES  (VARIABLES)

total_x = df.loc['suma', 'x']
print(f"El total de la columna 'x' es: {total_x}")
```

```python


total_y = df.loc['suma', 'y']
print(f"El total de la columna 'y' es: {total_y}")
```

```python
total_xy = df.loc['suma', 'xy']
print(f"El total de la columna 'xy' es: {total_xy}")
```

```python
total_x2 = df.loc['suma', 'x2']
print(f"El total de la columna 'x2' es: {total_x2}")
```

```python
# COMO DETERMINAR LA SUMA DE LAS FILAS

num_filas = df.shape[0]
print(f"El número total de filas en el DataFrame es: {num_filas}")
```

```python
num_filas_len = len(datos1)
print(f"El número total de filas en el DataFrame (usando len()) es: {num_filas_len}")
```

Ten en cuenta que `df.shape[0]` te dará el número total de filas incluyendo la fila 'suma' que añadiste. Si quieres el número de filas de datos originales (sin la fila 'suma'), tendrías que restar 1 o usar `len(df.drop('suma'))`.

```python
num_filas_datos_originales = len(df.drop('suma'))
print(f"El número de filas de datos originales (excluyendo la fila 'suma') es: {num_filas_datos_originales}")
```
